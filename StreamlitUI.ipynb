```python
%%writefile app.py
import streamlit as st
from ollama import chat

st.title("🎬 Movie Recommendation Assistant")
st.write("Ask me for a movie recommendation!")

if "messages" not in st.session_state:
    st.session_state.messages = []

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.write(msg["content"])

user_input = st.chat_input("What kind of movie are you looking for?")

if user_input:
    st.session_state.messages.append({"role": "user", "content": user_input})
    with st.chat_message("user"):
        st.write(user_input)

    response = chat(model="llama3.2:3b", messages=st.session_state.messages)
    reply = response["message"]["content"]

    st.session_state.messages.append({"role": "assistant", "content": reply})
    with st.chat_message("assistant"):
        st.write(reply)
```
